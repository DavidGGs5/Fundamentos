```python
base = int(input("Ingresa la base: "))
altura = int(input("Ingresa la altura: "))


def calcular_area(base, altura):
  area = base * altura
  print(area)

calcular_area(base, altura)
```

```python
x = int(input("Ingresa la tabla que quieres: "))
y = int(input("Hasta que numero debe llegar (no superior a 100): "))

def tablas_de_multiplicar(x, y):
  if y <= 100:
    for i in range(1, y + 1):
      print(x, "x", i , "=",x * i)
  else:
    print("La tabla supera el numero 100")

tablas_de_multiplicar(x,y)
```

```python
tablas_de_multiplicar(3)
print(" ")
tablas_de_multiplicar(5)
```
