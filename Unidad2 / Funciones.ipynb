```python
x = int(input("Ingresa la tabla que quieres: "))
y = int(input("Hasta que numero debe llegar (no superior a 100): "))

def tablas_de_multiplicar(x, y):
  if y <= 100:
    for i in range(1, y + 1):
      print(x, "x", i , "=",x * i)
  else:
    print("La tabla supera el numero 100")

tablas_de_multiplicar(x,y)
```

# **Alcance de las variables**

- El alcance de una variable se refiere a la parte del codigo donde se conoce la variable.

- Una variable que se define dentro de una función NO existe en otras funciones o fuera de la función.

- Para extender el alcance de una variable y que pueda existir fuera de una funcion se debe definir con la palabra reservada **global**.

```python
def miFuncion():
  # La variablex de miFuncion existe unicamente dentro del
  # codigo de la funcion
  global variablex
  variablex = 123
  print(variablex)

miFuncion()
# La siguiente variable NO es la misma que esta definida en MiFuncion,
# por lo tanto, no existe la variable y por eso marca error
print(variablex)
```

```python
def funcion2():
  variable1 = 5
  print("El valor de variable1 es: ", variable1)

variable1 = 10
print("Valor de la variable1 es: ", variable1)
funcion2()
print("Valor de la variable1 es: ", variable1)
```
