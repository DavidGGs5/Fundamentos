# **Tema de listas**

```python
alumnos = ["Ana", "Luis", "Carlos", "Juan"]
print(alumnos)
print(alumnos[0])
print(alumnos[3])
print(alumnos[5]) #Error de indice, el 5 no existe
```

```python
#Modificar un valor de la lista desde un indice especifico
alumnos[1] = "Maria Luisa"
print(alumnos)
```

```python
#1 Metodo append agrega un elemento al final de la lista
alumnos.append("Karla")
print(alumnos)
```

```python
#2 El metodo insert agrega un elemento a la lista en la posicion especificada
alumnos.insert(2, "Cristina")
print(alumnos)
```

```python
#3 El metodo remove, elimina un elemento especificando el valor del elemento
# debe coincidir exactamente mayusculas y minusculas
alumnos.remove("Ana")
alumnos.remove("ana") #Este marcaria error porque no esta bien escrito
print(alumnos)
```

```python
#4 pop elimina el elemento o item en la posicion especificada
alumnos.pop(2)
print(alumnos)
```

```python
#5 del tambien elimina el elemento o item en la posicion especificada
del alumnos[3]
print(alumnos)
```

```python
# La  funcion len me permite conocer cuantos elementos tiene una lista
print("Elementos en la lista de alumnos: ", len(alumnos))
```
